```python
from PIL import Image
import os
from collections import Counter

CURRENCY_PATH = "/kaggle/input/datasets/preetrank/indian-currency-real-vs-fake-notes-dataset/data/data"

sizes = Counter()
formats = Counter()
corrupted = []

for label in ["real", "fake"]:
    for denomination in os.listdir(os.path.join(CURRENCY_PATH, label)):
        folder = os.path.join(CURRENCY_PATH, label, denomination)

        if not os.path.isdir(folder):
            continue

        for filename in os.listdir(folder):
            if not filename.lower().endswith((".jpg", ".jpeg", ".png")):
                continue

            filepath = os.path.join(folder, filename)

            try:
                with Image.open(filepath) as img:
                    sizes[img.size] += 1
                    formats[img.format] += 1
                    img.verify()

            except Exception:
                corrupted.append(filepath)

print("Total images checked:", sum(sizes.values()))
print("\nMost common image sizes:")
for size, count in sizes.most_common(15):
    print(size, ":", count)

print("\nImage formats:")
for fmt, count in formats.items():
    print(fmt, ":", count)

print("\nCorrupted images:", len(corrupted))

if corrupted:
    print("\nFirst 10 corrupted files:")
    for f in corrupted[:10]:
        print(f)
```

```python
import os
import pandas as pd

CURRENCY_PATH = "/kaggle/input/datasets/preetrank/indian-currency-real-vs-fake-notes-dataset/data/data"

records = []

for label_name, label in [("real", 0), ("fake", 1)]:

    label_path = os.path.join(CURRENCY_PATH, label_name)

    for denomination in os.listdir(label_path):

        denomination_path = os.path.join(label_path, denomination)

        if not os.path.isdir(denomination_path):
            continue

        for filename in os.listdir(denomination_path):

            if filename.lower().endswith((".jpg", ".jpeg", ".png")):

                records.append({
                    "filepath": os.path.join(denomination_path, filename),
                    "label": label,
                    "class": label_name,
                    "denomination": denomination
                })

df = pd.DataFrame(records)

print("Total images:", len(df))
print("\nClass distribution:")
print(df["class"].value_counts())

print("\nDenomination distribution:")
print(df["denomination"].value_counts().sort_index())

print("\nFirst 5 records:")
display(df.head())
```

```python
from sklearn.model_selection import train_test_split

# Create a combined stratification group
df["stratify_group"] = (
    df["class"] + "_" + df["denomination"].astype(str)
)

# 70% Train, 30% Temporary
train_df, temp_df = train_test_split(
    df,
    test_size=0.30,
    random_state=42,
    stratify=df["stratify_group"]
)

# 15% Validation, 15% Test
val_df, test_df = train_test_split(
    temp_df,
    test_size=0.50,
    random_state=42,
    stratify=temp_df["stratify_group"]
)

print("Dataset split:")
print("------------------------")
print("Training   :", len(train_df))
print("Validation :", len(val_df))
print("Testing    :", len(test_df))
print("Total      :", len(train_df) + len(val_df) + len(test_df))

print("\nTraining class distribution:")
print(train_df["class"].value_counts())

print("\nValidation class distribution:")
print(val_df["class"].value_counts())

print("\nTesting class distribution:")
print(test_df["class"].value_counts())
```

```python
print("\nTraining denomination distribution:")
print(pd.crosstab(train_df["denomination"], train_df["class"]))

print("\nValidation denomination distribution:")
print(pd.crosstab(val_df["denomination"], val_df["class"]))

print("\nTesting denomination distribution:")
print(pd.crosstab(test_df["denomination"], test_df["class"]))
```

```python
import tensorflow as tf
import numpy as np

IMG_SIZE = (224, 224)
BATCH_SIZE = 32
AUTOTUNE = tf.data.AUTOTUNE

def load_image(filepath, label):
    image = tf.io.read_file(filepath)
    image = tf.image.decode_image(
        image,
        channels=3,
        expand_animations=False
    )

    image = tf.image.resize(image, IMG_SIZE)
    image = tf.cast(image, tf.float32) / 255.0

    return image, tf.cast(label, tf.float32)


def create_dataset(dataframe, shuffle=False):
    paths = dataframe["filepath"].values
    labels = dataframe["label"].values.astype(np.float32)

    dataset = tf.data.Dataset.from_tensor_slices((paths, labels))

    if shuffle:
        dataset = dataset.shuffle(
            buffer_size=len(dataframe),
            seed=42,
            reshuffle_each_iteration=True
        )

    dataset = dataset.map(
        load_image,
        num_parallel_calls=AUTOTUNE
    )

    dataset = dataset.batch(BATCH_SIZE)
    dataset = dataset.prefetch(AUTOTUNE)

    return dataset


train_ds = create_dataset(train_df, shuffle=True)
val_ds = create_dataset(val_df)
test_ds = create_dataset(test_df)

print("TensorFlow datasets created successfully.")
```

```python
import os
import pandas as pd

CURRENCY_PATH = "/kaggle/input/datasets/preetrank/indian-currency-real-vs-fake-notes-dataset/data/data"

records = []

for label_name, label in [("real", 0), ("fake", 1)]:

    label_path = os.path.join(CURRENCY_PATH, label_name)

    for denomination in os.listdir(label_path):

        denomination_path = os.path.join(label_path, denomination)

        if not os.path.isdir(denomination_path):
            continue

        for filename in os.listdir(denomination_path):

            if filename.lower().endswith((".jpg", ".jpeg", ".png")):

                records.append({
                    "filepath": os.path.join(denomination_path, filename),
                    "label": label,
                    "class": label_name,
                    "denomination": denomination
                })

df = pd.DataFrame(records)

print("Total images:", len(df))
print(df["class"].value_counts())
```

```python
from sklearn.model_selection import train_test_split

df["stratify_group"] = (
    df["class"] + "_" + df["denomination"].astype(str)
)

train_df, temp_df = train_test_split(
    df,
    test_size=0.30,
    random_state=42,
    stratify=df["stratify_group"]
)

val_df, test_df = train_test_split(
    temp_df,
    test_size=0.50,
    random_state=42,
    stratify=temp_df["stratify_group"]
)

print("Training   :", len(train_df))
print("Validation :", len(val_df))
print("Testing    :", len(test_df))
```

```python
from sklearn.utils.class_weight import compute_class_weight
import numpy as np

classes = np.array([0, 1])

class_weights_array = compute_class_weight(
    class_weight="balanced",
    classes=classes,
    y=train_df["label"].values
)

class_weights = {
    int(cls): float(weight)
    for cls, weight in zip(classes, class_weights_array)
}

print("Class weights:")
print("Real (0):", class_weights[0])
print("Fake (1):", class_weights[1])
```

```python
import tensorflow as tf
from tensorflow.keras import layers, models

# Data augmentation suitable for natural currency images
data_augmentation = tf.keras.Sequential([
    layers.RandomRotation(0.05),
    layers.RandomZoom(0.10),
    layers.RandomContrast(0.10),
    layers.RandomTranslation(
        height_factor=0.05,
        width_factor=0.05
    )
], name="currency_augmentation")


def build_custom_cnn(input_shape=(224, 224, 3)):

    inputs = layers.Input(shape=input_shape)

    # Data augmentation
    x = data_augmentation(inputs)

    # Block 1
    x = layers.Conv2D(32, (3, 3), padding="same")(x)
    x = layers.BatchNormalization()(x)
    x = layers.ReLU()(x)
    x = layers.MaxPooling2D((2, 2))(x)

    # Block 2
    x = layers.Conv2D(64, (3, 3), padding="same")(x)
    x = layers.BatchNormalization()(x)
    x = layers.ReLU()(x)
    x = layers.MaxPooling2D((2, 2))(x)

    # Block 3
    x = layers.Conv2D(128, (3, 3), padding="same")(x)
    x = layers.BatchNormalization()(x)
    x = layers.ReLU()(x)
    x = layers.MaxPooling2D((2, 2))(x)

    # Block 4
    x = layers.Conv2D(256, (3, 3), padding="same")(x)
    x = layers.BatchNormalization()(x)
    x = layers.ReLU()(x)

    # Global feature extraction
    x = layers.GlobalAveragePooling2D()(x)

    # Fully connected layer
    x = layers.Dense(128, activation="relu")(x)
    x = layers.Dropout(0.40)(x)

    # Binary classification
    outputs = layers.Dense(1, activation="sigmoid")(x)

    return models.Model(
        inputs,
        outputs,
        name="FinGuard_Custom_CNN"
    )


custom_cnn = build_custom_cnn()

custom_cnn.summary()
```

```python
from tensorflow.keras.optimizers import Adam

custom_cnn.compile(
    optimizer=Adam(learning_rate=0.001),
    loss="binary_crossentropy",
    metrics=[
        "accuracy",
        tf.keras.metrics.Precision(name="precision"),
        tf.keras.metrics.Recall(name="recall"),
        tf.keras.metrics.AUC(name="auc")
    ]
)

print("Custom CNN compiled successfully.")
```

```python
from tensorflow.keras.callbacks import (
    EarlyStopping,
    ReduceLROnPlateau,
    ModelCheckpoint
)

callbacks = [
    EarlyStopping(
        monitor="val_auc",
        mode="max",
        patience=5,
        restore_best_weights=True,
        verbose=1
    ),

    ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.5,
        patience=2,
        min_lr=1e-6,
        verbose=1
    ),

    ModelCheckpoint(
        "finguard_currency_custom_cnn.keras",
        monitor="val_auc",
        mode="max",
        save_best_only=True,
        verbose=1
    )
]

print("Callbacks configured successfully.")
```

```python
history = custom_cnn.fit(
    train_ds,
    validation_data=val_ds,
    epochs=30,
    class_weight=class_weights,
    callbacks=callbacks
)
```

```python
import numpy as np
from sklearn.metrics import (
    confusion_matrix,
    classification_report,
    roc_auc_score
)

# Get predictions
y_true = []
y_prob = []

for images, labels in test_ds:
    predictions = model.predict(images, verbose=0)

    y_true.extend(labels.numpy())
    y_prob.extend(predictions.ravel())

y_true = np.array(y_true)
y_prob = np.array(y_prob)

# Convert probabilities to classes
y_pred = (y_prob >= 0.5).astype(int)

# Confusion Matrix
cm = confusion_matrix(y_true, y_pred)

print("===== CONFUSION MATRIX =====")
print(cm)

print("\n===== CLASSIFICATION REPORT =====")
print(classification_report(
    y_true,
    y_pred,
    target_names=["Real", "Fake"],
    digits=4
))

print("\n===== ROC-AUC =====")
print(f"{roc_auc_score(y_true, y_prob):.4f}")
```
